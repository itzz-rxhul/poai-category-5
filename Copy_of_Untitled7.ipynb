```python
#CATEGORY-5: 1.Greedy Best-First Search — 8 Puzzle
import heapq
def manhattan(state, goal):
    distance = 0

    for i in range(9):
        if state[i] != 0:
            current_row, current_col = divmod(i, 3)
            goal_index = goal.index(state[i])
            goal_row, goal_col = divmod(goal_index, 3)

            distance += abs(current_row - goal_row)
            distance += abs(current_col - goal_col)

    return distance
def get_neighbors(state):
    neighbors = []

    zero = state.index(0)
    row, col = divmod(zero, 3)

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_zero = new_row * 3 + new_col

            new_state = list(state)
            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            neighbors.append(tuple(new_state))

    return neighbors
def greedy_bfs(start, goal):
    priority_queue = []
    heapq.heappush(priority_queue, (manhattan(start, goal), start))

    visited = set()

    while priority_queue:
        h, current = heapq.heappop(priority_queue)

        if current in visited:
            continue

        visited.add(current)

        print("Visiting:", current, "Heuristic:", h)

        if current == goal:
            print("Goal reached!")
            return

        for neighbor in get_neighbors(current):
            if neighbor not in visited:
                heapq.heappush(
                    priority_queue,
                    (manhattan(neighbor, goal), neighbor)
                )

    print("No solution found.")
start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

greedy_bfs(start, goal)
```

```python
#2.A* Search — 8 Puzzle
import heapq

def manhattan(state, goal):
    distance = 0

    for i in range(9):
        if state[i] != 0:
            row, col = divmod(i, 3)

            goal_index = goal.index(state[i])
            goal_row, goal_col = divmod(goal_index, 3)

            distance += abs(row - goal_row)
            distance += abs(col - goal_col)

    return distance

def get_neighbors(state):
    neighbors = []

    zero = state.index(0)
    row, col = divmod(zero, 3)

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_zero = new_row * 3 + new_col

            new_state = list(state)

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            neighbors.append(tuple(new_state))

    return neighbors
def a_star(start, goal):

    priority_queue = []

    g_cost = {start: 0}

    f_cost = manhattan(start, goal)

    heapq.heappush(priority_queue, (f_cost, start))

    visited = set()

    while priority_queue:

        f, current = heapq.heappop(priority_queue)

        if current in visited:
            continue

        visited.add(current)

        print("State:", current)
        print("f(n) =", f)

        if current == goal:
            print("Goal reached!")
            print("Cost =", g_cost[current])
            return

        for neighbor in get_neighbors(current):

            new_g = g_cost[current] + 1

            if neighbor not in g_cost or new_g < g_cost[neighbor]:

                g_cost[neighbor] = new_g

                h = manhattan(neighbor, goal)
                f = new_g + h

                heapq.heappush(priority_queue, (f, neighbor))

    print("No solution found.")

start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

a_star(start, goal)
```

```python
#3.AO* Search
class AOStar:

    def __init__(self, graph, heuristic):
        self.graph = graph
        self.heuristic = heuristic
        self.solution = []

    def ao_star(self, node):

        print("Expanding:", node)

        # Goal node
        if node not in self.graph:
            return self.heuristic[node]

        # Find the best option
        best_cost = float('inf')
        best_option = None

        for option in self.graph[node]:

            # AND nodes
            cost = 0

            for child in option:
                cost += self.ao_star(child)

            if cost < best_cost:
                best_cost = cost
                best_option = option

        self.heuristic[node] = best_cost

        self.solution.append((node, best_option))

        return best_cost
graph = {

    'A': [['B', 'C'], ['D']],

    'B': [['E'], ['F']],

    'C': [['G']],

    'D': [['H', 'I']]
}
heuristic = {

    'A': 10,
    'B': 4,
    'C': 2,
    'D': 5,
    'E': 1,
    'F': 2,
    'G': 1,
    'H': 2,
    'I': 2
}
ao = AOStar(graph, heuristic)

cost = ao.ao_star('A')

print("\nMinimum Cost:", cost)

print("\nSolution Graph:")

for node, option in ao.solution:
    print(node, "->", option)
```

```python
#4.Memory-Bounded Heuristic Search
def manhattan(state, goal):

    distance = 0

    for i in range(9):

        if state[i] != 0:

            row, col = divmod(i, 3)

            goal_index = goal.index(state[i])
            goal_row, goal_col = divmod(goal_index, 3)

            distance += abs(row - goal_row)
            distance += abs(col - goal_col)

    return distance
def get_neighbors(state):

    neighbors = []

    zero = state.index(0)

    row, col = divmod(zero, 3)

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_zero = new_row * 3 + new_col

            new_state = list(state)

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            neighbors.append(tuple(new_state))

    return neighbors
def search(path, goal, g, threshold):

    current = path[-1]

    f = g + manhattan(current, goal)

    if f > threshold:
        return f

    if current == goal:
        return "FOUND"

    minimum = float('inf')

    for neighbor in get_neighbors(current):

        if neighbor not in path:

            path.append(neighbor)

            result = search(
                path,
                goal,
                g + 1,
                threshold
            )

            if result == "FOUND":
                return "FOUND"

            if result < minimum:
                minimum = result

            path.pop()

    return minimum
def ida_star(start, goal):

    threshold = manhattan(start, goal)

    path = [start]

    while True:

        print("Current threshold:", threshold)

        result = search(
            path,
            goal,
            0,
            threshold
        )

        if result == "FOUND":

            print("\nGoal reached!")

            print("\nSolution Path:")

            for state in path:
                print(state)

            return

        if result == float('inf'):

            print("No solution found.")
            return

        threshold = result
start = (
    1, 2, 3,
    4, 0, 6,
    7, 5, 8
)

goal = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)

ida_star(start, goal)
```
